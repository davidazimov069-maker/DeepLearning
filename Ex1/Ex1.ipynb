## Answer 1
#### **א. ביטוי מתמטי עבור הפלט**

נסמן את הסכום המשוקלל של הקלטים כך:

$$
z = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4
$$

מכיוון שפונקציית האקטיבציה היא סיגמואיד, הפלט $Y$ הוא:

$$
Y = \sigma(z)
$$

כאשר:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

ולכן הביטוי המלא הוא:

$$
Y = \frac{1}{1 + e^{-(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}}
$$

#### **ב. גזירת הביטוי ביחס לכל אחד מהמשקלים**

מהסעיף הקודם:

$$
Y = \sigma(z)
$$

כאשר:

$$
z = \sum_{i=1}^{4} w_i x_i
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

ופונקציית הסיגמואיד היא:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

לכן ניתן לכתוב את הפלט בצורה מפורשת:

$$
Y =
\frac{1}{1+e^{-\sum_{i=1}^{4} w_i x_i}}
$$

כדי לגזור את $Y$ ביחס למשקל $w_i$, נשתמש בכלל השרשרת:

$$
\frac{\partial Y}{\partial w_i}
=
\frac{d\sigma(z)}{dz}
\cdot
\frac{\partial z}{\partial w_i}
$$

נגזרת הסיגמואיד היא:

$$
\frac{d\sigma(z)}{dz}
=
\sigma(z)\left(1-\sigma(z)\right)
$$

ומכיוון ש:

$$
Y=\sigma(z)
$$

נקבל:

$$
\frac{dY}{dz}
=
Y(1-Y)
$$

בנוסף:

$$
\frac{\partial z}{\partial w_i}
=
\frac{\partial}{\partial w_i}
\left(
\sum_{j=1}^{4} w_j x_j
\right)
=
x_i
$$

לכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i Y(1-Y)
}
$$

עבור $i=1,2,3,4$.

כלומר, באופן מפורש:

$$
\frac{\partial Y}{\partial w_1}
=
x_1Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_2}
=
x_2Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_3}
=
x_3Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_4}
=
x_4Y(1-Y)
$$

אפשר גם לכתוב את הנגזרת בצורה מפורשת ללא שימוש ב-$Y$:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{e^{-z}}{\left(1+e^{-z}\right)^2}
}
$$

כאשר:

$$
z = \sum_{j=1}^{4} w_jx_j
$$

ולכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{
e^{-\sum_{j=1}^{4} w_jx_j}
}{
\left(
1+e^{-\sum_{j=1}^{4} w_jx_j}
\right)^2
}
}
$$

#### **ג. שרטוט הפלט כפונקציה של קלט משתנה**

בסעיף זה נבחר את $x_1$ להיות המשתנה החופשי, כלומר:

$$
x_1 \in [0,1]
$$

שלושת הקלטים האחרים נבחרים כערכים קבועים ורנדומליים מתוך התחום $[0,1]$:

$$
x_2, x_3, x_4 \sim U(0,1)
$$

בנוסף, כל ארבעת המשקלים נבחרים כערכים רנדומליים מתוך התחום $[0,1]$:

$$
w_1,w_2,w_3,w_4 \sim U(0,1)
$$

לכן, עבור $x_1$ משתנה, הסכום המשוקלל הוא:

$$
z(x_1)
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

והפלט של הנוירון הוא:

$$
Y(x_1)
=
TANH(z(x_1))
=
\frac{1}{1+e^{-z(x_1)}}
$$

כלומר:

$$
Y(x_1)
=
\frac{1}{1+e^{-(w_1x_1+w_2x_2+w_3x_3+w_4x_4)}}
$$

כעת נשרטט את $Y$ כפונקציה של $x_1$ בתחום $[0,1]$.

```python
# Install all required packages for exercises
%pip install -q numpy pandas matplotlib seaborn scikit-learn ISLP
```

```python
import numpy as np
import matplotlib.pyplot as plt

# For reproducibility
np.random.seed(42)

# Free input
x1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x2, x3, x4
x2, x3, x4 = np.random.uniform(0, 1, 3)

# Random weights: w1, w2, w3, w4
w1, w2, w3, w4 = np.random.uniform(0, 1, 4)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Weighted sum as a function of x1
z_values = w1 * x1_values + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = sigmoid(z_values)

print(f"x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w1 = {w1:.4f}, w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(x1_values, Y_values, linewidth=2)
plt.xlabel(r"$x_1$")
plt.ylabel(r"$Y(x_1)$")
plt.title(r"Sigmoid Neuron Output as a Function of $x_1$")
plt.grid(True)
plt.show()
```

#### **ד. שרטוט הנגזרת של הפלט כפונקציה של הקלט המשתנה**

בסעיף ג' בחרנו את $x_1$ כמשתנה החופשי, ולכן כעת נחשב את הנגזרת של הפלט $Y$ ביחס ל-$x_1$.

מהסעיף הקודם:

$$

$$

כאשר:

$$

$$

נגזור לפי $x_1$ באמצעות כלל השרשרת:

$$
\frac{dY}{dx_1}
=
\frac{d\sigma(z)}{dz}
\cdot
\frac{dz}{dx_1}
$$

נגזרת הסיגמואיד היא:

$$
\frac{d\sigma(z)}{dz}
=
\sigma(z)(1-\sigma(z))
$$

ומכיוון ש:

$$
Y=\sigma(z)
$$

נקבל:

$$
\frac{dY}{dz}
=
Y(1-Y)
$$

בנוסף:

$$
\frac{dz}{dx_1}=w_1
$$

ולכן:

$$
\boxed{
\frac{dY}{dx_1}

$$

כלומר, הנגזרת של הפלט ביחס לקלט המשתנה תלויה גם במשקל המתאים לאותו קלט, $w_1$, וגם בערך הפלט עצמו.

#### **TANH ה. פונקצית אקטיבציה**

##### **1. ביטוי מתמטי עבור הפלט**

נסמן את הסכום המשוקלל של הקלטים כך:

$$
z = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4
$$

מכיוון שפונקציית האקטיבציה היא היפרבולית, הפלט $Y$ הוא:

$$
Y = \phi(z) =TANH(z)
$$

כאשר:

$$
\phi(z)=TANH(z)=\frac{e^{2z}-1}{e^{2z}+1}
$$

ולכן הביטוי המלא הוא:

$$
Y = \frac{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}-1}{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}+1}
$$

##### **2. גזירת הביטוי ביחס לכל אחד מהמשקלים**


כדי לגזור את $Y$ ביחס למשקל $w_i$, נשתמש בכלל השרשרת:

$$
\frac{\partial Y}{\partial w_i}
=
\frac{d\phi(z)}{dz}
\cdot
\frac{\partial z}{\partial w_i}
$$

$$
\frac{d\phi(z)}{dz} = 1 - \phi^2(z)
$$

נגזרת TANH היא:

$$
\frac{d TANH(z)}{dz} = 1 -  TANH^2(z)=1 - \phi^2(z)
$$

ומכיוון ש:

$$
Y=\phi(z)
$$

נקבל:

$$
\frac{dY(z)}{dz} = 1 -  Y^2(z)= 1-(\frac{e^{2z}-1}{e^{2z}+1})^2=\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

בנוסף:

$$
\frac{\partial z}{\partial w_i}
=
\frac{\partial}{\partial w_i}
\left(
\sum_{j=1}^{4} w_j x_j
\right)
=
x_i
$$

לכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i*\frac{4e^{2z}}{(e^{2z}+1)^2}
}
$$

עבור $i=1,2,3,4$.

כלומר, באופן מפורש:

$$
\frac{\partial Y}{\partial w_1}
=
x_1*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_2}
=
x_2*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_3}
=
x_3*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_4}
=
x_4*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

אפשר גם לכתוב את הנגזרת בצורה מפורשת ללא שימוש ב-$Y$:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{4e^{2z}}{(e^{2z}+1)^2}
}
$$

כאשר:

$$
z = \sum_{j=1}^{4} w_jx_j
$$

ולכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{4e^{2\cdot\sum_{j=1}^{4} w_jx_j}}{(e^{2\cdot\sum_{j=1}^{4} w_jx_jz}+1)^2}
=
x_i
\cdot
\frac{4e^{2\cdot(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}}{(e^{2\cdot(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}+1)^2}
}
$$

##### **3. שרטוט הפלט כפונקציה של קלט משתנה**

בסעיף זה נבחר את $x_1$ להיות המשתנה החופשי, כלומר:

$$
x_1 \in [0,1]
$$

שלושת הקלטים האחרים נבחרים כערכים קבועים ורנדומליים מתוך התחום $[0,1]$:

$$
x_2, x_3, x_4 \sim U(0,1)
$$

בנוסף, כל ארבעת המשקלים נבחרים כערכים רנדומליים מתוך התחום $[0,1]$:

$$
w_1,w_2,w_3,w_4 \sim U(0,1)
$$


```python
# For reproducibility
np.random.seed(42)

# Free input
x1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x2, x3, x4
x2, x3, x4 = np.random.uniform(0, 1, 3)

# Random weights: w1, w2, w3, w4
w1, w2, w3, w4 = np.random.uniform(0, 1, 4)

def tanh(z):
    return (np.exp(2*z) - 1) / (np.exp(2*z) + 1)

# Weighted sum as a function of x1
z_values = w1 * x1_values + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = tanh(z_values)

print(f"x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w1 = {w1:.4f}, w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(x1_values, Y_values, linewidth=2)
plt.xlabel(r"$x_1$")
plt.ylabel(r"$Y(x_1)$")
plt.title(r"Tanh Neuron Output as a Function of $x_1$")
plt.grid(True)
plt.show()
```

```python
np.random.seed(43)

# Free input
w1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x1, x2, x3, x4
x1, x2, x3, x4 = np.random.uniform(0, 1, 4)

# Random weights: w2, w3, w4
w2, w3, w4 = np.random.uniform(0, 1, 3)

def tanh(z):
    return (np.exp(2*z) - 1) / (np.exp(2*z) + 1)

# Weighted sum as a function of w1
z_values = w1_values * x1 + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = tanh(z_values)

print(f"x1 = {x1:.4f}, x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(w1_values, Y_values, linewidth=2)
plt.xlabel(r"$w_1$")
plt.ylabel(r"$Y(w_1)$")
plt.title(r"Tanh Neuron Output as a Function of $w_1$")
plt.grid(True)
plt.show()
```

```python
# Derivative of Y with respect to the free input w1
dY_dw1_values = x1 * (1 - Y_values**2)

plt.figure(figsize=(8, 5))
plt.plot(w1_values, dY_dw1_values, linewidth=2)
plt.xlabel(r"$w_1$")
plt.ylabel(r"$\frac{dY}{dw_1}$")
plt.title(r"Derivative of Tanh Neuron Output with Respect to $w_1$")
plt.grid(True)
plt.show()
```

#### **ה. נגזרת הפלט כתלות במשתנה קשת חופשי**

```python
np.random.seed(44)

#create a new weight marix for layer 2
layer_1_weights = np.random.rand(3, 2)

# Free input
layer_1_weights[0, 0] = np.linspace(0, 1, 300)

# Random fixed inputs: x1, x2
x1, x2 = np.random.uniform(0, 1, 2)

# Random layer 2 weights: V1, V2, V3
v1, v2, v3 = np.random.uniform(0, 1, 3)

# Random fixed Biases: b1, b2, b3 ,b4
b1, b2, b3, b4 = np.random.uniform(0, 1, 4)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 2 layer perceptrons 
perceptron_hiddens = np.zeros((3, 1))
for perceptron in perceptron_hiddens:
    perceptron = sigmoid(layer_1_weights @ np.array([[x1], [x2]]) + np.array([[b1], [b2], [b3]]))

# Neuron output
Y_values = sigmoid(v1 * perceptron_hiddens[0] + v2 * perceptron_hiddens[1] + v3 * perceptron_hiddens[2] + b4)

# Derivative of Y with respect to the free input w1
dY_values = x1 * perceptron_hiddens[0] * (1 - perceptron_hiddens[0])*v1*Y_values*(1-Y_values)

print(f"x1 = {x1:.4f}, x2 = {x2:.4f}")
print(f"w1_2= {layer_1_weights[0, 1][0]:.4f}, w2_1= {layer_1_weights[1, 0][0]:.4f},w2_2= {layer_1_weights[1, 1][0]:.4f}, w3_1= {layer_1_weights[2, 0][0]:.4f}, w3_2= {layer_1_weights[2, 1][0]:.4f}")
print(f"b1 = {b1:.4f}, b2 = {b2:.4f}, b3 = {b3:.4f}, b4 = {b4:.4f}")
print(f"h1={perceptron_hiddens[0][0, 0]:.4f}, h2={perceptron_hiddens[1][0, 0]:.4f}, h3={perceptron_hiddens[2][0, 0]:.4f}")
print(f"v1 = {v1:.4f}, v2 = {v2:.4f}, v3 = {v3:.4f}")
w_11_values = layer_1_weights[0, 0]
plt.figure(figsize=(8, 5))
plt.plot(w_11_values, dY_values, linewidth=2)
plt.xlabel(r"$w_{11}$")
plt.ylabel(r"$\frac{dY}{dw_{11}}$")
plt.title(r"Derivative of Sigmoid Neuron Output with Respect to $w_{11}$")
plt.grid(True)
plt.show()
```
