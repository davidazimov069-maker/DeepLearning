## Answer 1
#### **א. ביטוי מתמטי עבור הפלט**

נסמן את הסכום המשוקלל של הקלטים כך:

$$
z = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4
$$

מכיוון שפונקציית האקטיבציה היא סיגמואיד, הפלט $Y$ הוא:

$$
Y = \sigma(z)
$$

כאשר:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

ולכן הביטוי המלא הוא:

$$
Y = \frac{1}{1 + e^{-(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}}
$$

#### **ב. גזירת הביטוי ביחס לכל אחד מהמשקלים**

מהסעיף הקודם:

$$
Y = \sigma(z)
$$

כאשר:

$$
z = \sum_{i=1}^{4} w_i x_i
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

ופונקציית הסיגמואיד היא:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

לכן ניתן לכתוב את הפלט בצורה מפורשת:

$$
Y =
\frac{1}{1+e^{-\sum_{i=1}^{4} w_i x_i}}
$$

כדי לגזור את $Y$ ביחס למשקל $w_i$, נשתמש בכלל השרשרת:

$$
\frac{\partial Y}{\partial w_i}
=
\frac{d\sigma(z)}{dz}
\cdot
\frac{\partial z}{\partial w_i}
$$

נגזרת הסיגמואיד היא:

$$
\frac{d\sigma(z)}{dz}
=
\sigma(z)\left(1-\sigma(z)\right)
$$

ומכיוון ש:

$$
Y=\sigma(z)
$$

נקבל:

$$
\frac{dY}{dz}
=
Y(1-Y)
$$

בנוסף:

$$
\frac{\partial z}{\partial w_i}
=
\frac{\partial}{\partial w_i}
\left(
\sum_{j=1}^{4} w_j x_j
\right)
=
x_i
$$

לכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i Y(1-Y)
}
$$

עבור $i=1,2,3,4$.

כלומר, באופן מפורש:

$$
\frac{\partial Y}{\partial w_1}
=
x_1Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_2}
=
x_2Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_3}
=
x_3Y(1-Y)
$$

$$
\frac{\partial Y}{\partial w_4}
=
x_4Y(1-Y)
$$

אפשר גם לכתוב את הנגזרת בצורה מפורשת ללא שימוש ב-$Y$:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{e^{-z}}{\left(1+e^{-z}\right)^2}
}
$$

כאשר:

$$
z = \sum_{j=1}^{4} w_jx_j
$$

ולכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{
e^{-\sum_{j=1}^{4} w_jx_j}
}{
\left(
1+e^{-\sum_{j=1}^{4} w_jx_j}
\right)^2
}
}
$$

#### **ג. שרטוט הפלט כפונקציה של קלט משתנה**

בסעיף זה נבחר את $x_1$ להיות המשתנה החופשי, כלומר:

$$
x_1 \in [0,1]
$$

שלושת הקלטים האחרים נבחרים כערכים קבועים ורנדומליים מתוך התחום $[0,1]$:

$$
x_2, x_3, x_4 \sim U(0,1)
$$

בנוסף, כל ארבעת המשקלים נבחרים כערכים רנדומליים מתוך התחום $[0,1]$:

$$
w_1,w_2,w_3,w_4 \sim U(0,1)
$$

לכן, עבור $x_1$ משתנה, הסכום המשוקלל הוא:

$$
z(x_1)
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

והפלט של הנוירון הוא:

$$
Y(x_1)
=
TANH(z(x_1))
=
\frac{1}{1+e^{-z(x_1)}}
$$

כלומר:

$$
Y(x_1)
=
\frac{1}{1+e^{-(w_1x_1+w_2x_2+w_3x_3+w_4x_4)}}
$$

כעת נשרטט את $Y$ כפונקציה של $x_1$ בתחום $[0,1]$.

```python
# Install all required packages for exercises
%pip install -q numpy pandas matplotlib seaborn scikit-learn ISLP
```

```python
import numpy as np
import matplotlib.pyplot as plt

# For reproducibility
np.random.seed(42)

# Free input
x1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x2, x3, x4
x2, x3, x4 = np.random.uniform(0, 1, 3)

# Random weights: w1, w2, w3, w4
w1, w2, w3, w4 = np.random.uniform(0, 1, 4)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Weighted sum as a function of x1
z_values = w1 * x1_values + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = sigmoid(z_values)

print(f"x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w1 = {w1:.4f}, w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(x1_values, Y_values, linewidth=2)
plt.xlabel(r"$x_1$")
plt.ylabel(r"$Y(x_1)$")
plt.title(r"Sigmoid Neuron Output as a Function of $x_1$")
plt.grid(True)
plt.show()
```

#### **ד. שרטוט הנגזרת של הפלט כפונקציה של הקלט המשתנה**

בסעיף ג' בחרנו את $x_1$ כמשתנה החופשי, ולכן כעת נחשב את הנגזרת של הפלט $Y$ ביחס ל-$x_1$.

מהסעיף הקודם:

$$

$$

כאשר:

$$

$$

נגזור לפי $x_1$ באמצעות כלל השרשרת:

$$
\frac{dY}{dx_1}
=
\frac{d\sigma(z)}{dz}
\cdot
\frac{dz}{dx_1}
$$

נגזרת הסיגמואיד היא:

$$
\frac{d\sigma(z)}{dz}
=
\sigma(z)(1-\sigma(z))
$$

ומכיוון ש:

$$
Y=\sigma(z)
$$

נקבל:

$$
\frac{dY}{dz}
=
Y(1-Y)
$$

בנוסף:

$$
\frac{dz}{dx_1}=w_1
$$

ולכן:

$$
\boxed{
\frac{dY}{dx_1}

$$

כלומר, הנגזרת של הפלט ביחס לקלט המשתנה תלויה גם במשקל המתאים לאותו קלט, $w_1$, וגם בערך הפלט עצמו.

#### **TANH ה. פונקצית אקטיבציה**

##### **1. ביטוי מתמטי עבור הפלט**

נסמן את הסכום המשוקלל של הקלטים כך:

$$
z = w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4
$$

מכיוון שפונקציית האקטיבציה היא היפרבולית, הפלט $Y$ הוא:

$$
Y = \phi(z) =TANH(z)
$$

כאשר:

$$
\phi(z)=TANH(z)=\frac{e^{2z}-1}{e^{2z}+1}
$$

ולכן הביטוי המלא הוא:

$$
Y = \frac{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}-1}{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}+1}
$$

##### **2. גזירת הביטוי ביחס לכל אחד מהמשקלים**


כדי לגזור את $Y$ ביחס למשקל $w_i$, נשתמש בכלל השרשרת:

$$
\frac{\partial Y}{\partial w_i}
=
\frac{d\phi(z)}{dz}
\cdot
\frac{\partial z}{\partial w_i}
$$

$$
\frac{d\phi(z)}{dz} = 1 - \phi^2(z)
$$

נגזרת TANH היא:

$$
\frac{d TANH(z)}{dz} = 1 -  TANH^2(z)=1 - \phi^2(z)
$$

ומכיוון ש:

$$
Y=\phi(z)
$$

נקבל:

$$
\frac{dY(z)}{dz} = 1 -  Y^2(z)= 1-(\frac{e^{2z}-1}{e^{2z}+1})^2=\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

בנוסף:

$$
\frac{\partial z}{\partial w_i}
=
\frac{\partial}{\partial w_i}
\left(
\sum_{j=1}^{4} w_j x_j
\right)
=
x_i
$$

לכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i*\frac{4e^{2z}}{(e^{2z}+1)^2}
}
$$

עבור $i=1,2,3,4$.

כלומר, באופן מפורש:

$$
\frac{\partial Y}{\partial w_1}
=
x_1*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_2}
=
x_2*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_3}
=
x_3*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

$$
\frac{\partial Y}{\partial w_4}
=
x_4*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

אפשר גם לכתוב את הנגזרת בצורה מפורשת ללא שימוש ב-$Y$:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{4e^{2z}}{(e^{2z}+1)^2}
}
$$

כאשר:

$$
z = \sum_{j=1}^{4} w_jx_j
$$

ולכן:

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{4e^{2\cdot\sum_{j=1}^{4} w_jx_j}}{(e^{2\cdot\sum_{j=1}^{4} w_jx_jz}+1)^2}
=
x_i
\cdot
\frac{4e^{2\cdot(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}}{(e^{2\cdot(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}+1)^2}
}
$$

##### **3. שרטוט הפלט כפונקציה של קלט משתנה**

בסעיף זה נבחר את $x_1$ להיות המשתנה החופשי, כלומר:

$$
x_1 \in [0,1]
$$

שלושת הקלטים האחרים נבחרים כערכים קבועים ורנדומליים מתוך התחום $[0,1]$:

$$
x_2, x_3, x_4 \sim U(0,1)
$$

בנוסף, כל ארבעת המשקלים נבחרים כערכים רנדומליים מתוך התחום $[0,1]$:

$$
w_1,w_2,w_3,w_4 \sim U(0,1)
$$


```python
# For reproducibility
np.random.seed(42)

# Free input
x1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x2, x3, x4
x2, x3, x4 = np.random.uniform(0, 1, 3)

# Random weights: w1, w2, w3, w4
w1, w2, w3, w4 = np.random.uniform(0, 1, 4)

def tanh(z):
    return (np.exp(2*z) - 1) / (np.exp(2*z) + 1)

# Weighted sum as a function of x1
z_values = w1 * x1_values + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = tanh(z_values)

print(f"x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w1 = {w1:.4f}, w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(x1_values, Y_values, linewidth=2)
plt.xlabel(r"$x_1$")
plt.ylabel(r"$Y(x_1)$")
plt.title(r"Tanh Neuron Output as a Function of $x_1$")
plt.grid(True)
plt.show()
```

#### **א. שינוי מספר אפוקים**
בסעיף זה נראה כיצד שינוי מספר האפוקים משפיע על המודל
מספר אפוקים מייצג את כמות הפעמים שכל נתוני
האימון עברו דרך רשת הנוירונים
(מהקלט אל הפלט, תוך עדכון המשקולות).

מבחינה תיאורטית, ככל שמספר האפוקים נמוך
נקבל underfit:
המודל לא יספיק לראות את הנתונים 
מספיק פעמים כדי לכוונן את המשקולות
.כתוצאה מכך, השגיאה תישאר גבוהה והדיוק יהיה נמוך

ככל שמספר האפוקים גבוהה
נקבל overfit:
המודל עלול להתחיל "לשנן" את נתוני האימון
.הקיימים במקום ללמוד חוקיות כללית
במצב זה המודל יעבוד מעולה על דאטה האימון, אך ייכשל כשיראה נתונים חדשים.
.כתוצאה מכך, השגיאה תהיה גבוהה והדיוק יהיה נמוך עבור מידע חדש

```python
## edited MYFIRSTDEEPLEARNINGPROGRAM - Neural Network Implementation

import numpy as np
import matplotlib.pyplot as plt


class SimpleNeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size):
        # Random weight initialization
        self.weights_input_hidden = np.random.randn(input_size, hidden_size)
        self.bias_hidden = np.zeros((1, hidden_size))
        
        self.weights_hidden_output = np.random.randn(hidden_size, output_size)
        self.bias_output = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward(self, X):
        # Hidden layer
        self.hidden_layer = self.sigmoid(np.dot(X, self.weights_input_hidden) + self.bias_hidden)
        
        # Output layer
        self.output = self.sigmoid(np.dot(self.hidden_layer, self.weights_hidden_output) + self.bias_output)
        
        return self.output

    def train(self, X, y, epochs=10000, learning_rate=0.1):
        for _ in range(epochs):
            # Forward propagation
            output = self.forward(X)
            
            # Backpropagation
            output_error = y - output
            output_delta = output_error * self.sigmoid_derivative(output)
            
            hidden_error = np.dot(output_delta, self.weights_hidden_output.T)
            hidden_delta = hidden_error * self.sigmoid_derivative(self.hidden_layer)
            
            # Weight updates
            self.weights_hidden_output += learning_rate * np.dot(self.hidden_layer.T, output_delta)
            self.bias_output += learning_rate * np.sum(output_delta, axis=0, keepdims=True)
            
            self.weights_input_hidden += learning_rate * np.dot(X.T, hidden_delta)
            self.bias_hidden += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)

# Example usage
if __name__ == "__main__":
    # XOR problem dataset
    X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
    y = np.array([[0], [1], [1], [0]])
    EPOCHS = [10,100,1000,10000]
    error_values = []
    accuracy_values = []
    for epochs in EPOCHS:
        # Create and train neural network
        nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)
        nn.train(X, y, epochs=epochs)
        # Test predictions
        predictions = nn.forward(X)
        error = np.mean(np.abs(predictions - y))
        accuracy = 1 - error
        error_values.append(error)
        accuracy_values.append(accuracy)
        print(f"Predictions after {epochs} epochs vs Target Values:")
        for i in range(len(predictions)):
            print(f"Predicted: {predictions[i][0]:.4f} - Target: {y[i][0]}")
        print(f"error after {epochs} epochs: {error:.4f}")
        print(f"accuracy after {epochs} epochs: {accuracy:.4f}\n")

# #plot error and accuracy as a function of epochs
plt.figure(figsize=(10, 6))
plt.plot(EPOCHS, error_values, marker='o', color='red', linewidth=2, label='Error')
plt.plot(EPOCHS, accuracy_values, marker='s', color='blue', linewidth=2, label='Accuracy')
plt.xscale('log')
plt.ylim(-0.05, 1.05)
plt.title('Neural Network Performance: Error & Accuracy vs. Epochs')
plt.xlabel('Epochs (Logarithmic Scale)')
plt.ylabel('Value [0 to 1]')
plt.legend()
plt.grid(True, which="both", ls="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

השיפור בשגיאה ביחס לריצה הקודמת חייב להיות קטנה מ 0.001  

```python
class SimpleNeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size):
        # Random weight initialization
        self.weights_input_hidden = np.random.randn(input_size, hidden_size)
        self.bias_hidden = np.zeros((1, hidden_size))
        
        self.weights_hidden_output = np.random.randn(hidden_size, output_size)
        self.bias_output = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward(self, X):
        # Hidden layer
        self.hidden_layer = self.sigmoid(np.dot(X, self.weights_input_hidden) + self.bias_hidden)
        
        # Output layer
        self.output = self.sigmoid(np.dot(self.hidden_layer, self.weights_hidden_output) + self.bias_output)
        
        return self.output

    def train(self, X, y, epochs=10000, learning_rate=0.1):
        for _ in range(epochs):
            # Forward propagation
            output = self.forward(X)
            
            # Backpropagation
            output_error = y - output
            output_delta = output_error * self.sigmoid_derivative(output)
            
            hidden_error = np.dot(output_delta, self.weights_hidden_output.T)
            hidden_delta = hidden_error * self.sigmoid_derivative(self.hidden_layer)
            
            # Weight updates
            self.weights_hidden_output += learning_rate * np.dot(self.hidden_layer.T, output_delta)
            self.bias_output += learning_rate * np.sum(output_delta, axis=0, keepdims=True)
            
            self.weights_input_hidden += learning_rate * np.dot(X.T, hidden_delta)
            self.bias_hidden += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)

# Example usage
if __name__ == "__main__":
    # XOR problem dataset
    X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
    y = np.array([[0], [1], [1], [0]])
    # Create a large and detailed list of epochs (from 100 to 20,000 in steps of 500)
    EPOCHS = list(range(100, 20001, 500))
    
    error_values = []
    accuracy_values = []
    
    # Variables to track the optimal epoch (early stopping)
    best_epoch = 0
    min_error = float('inf')
    best_predictions = None
    
    # NEW: Variables to track the absolute best results across ALL epochs
    absolute_min_error = float('inf')
    absolute_max_accuracy = 0
    absolute_best_epoch = 0
    
    # Flag to lock the optimal epoch once the criteria are met
    optimal_found = False

    for epochs in EPOCHS:
        # Set the random seed so the comparison between epochs is perfectly fair
        np.random.seed(42) 
        
        # Create and train neural network
        nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)
        nn.train(X, y, epochs=epochs)
        
        # Test predictions
        predictions = nn.forward(X)
        error = np.mean(np.abs(predictions - y))
        accuracy = 1 - error
        
        error_values.append(error)
        accuracy_values.append(accuracy)
        
        # NEW: Track the absolute best values regardless of early stopping
        if error < absolute_min_error:
            absolute_min_error = error
            absolute_max_accuracy = accuracy
            absolute_best_epoch = epochs
            
        # Check for early stopping condition:
        # 1. Error must be strictly below 0.05 (5%)
        # 2. Improvement in error is less than 0.001
        if not optimal_found:
            if min_error < 0.05 and (min_error - error) < 0.001:
                optimal_found = True
            else:
                # Update best variables if we haven't locked them yet
                if error < min_error:
                    min_error = error
                    best_epoch = epochs
                    best_predictions = predictions

    # Print the results for the optimal number of epochs (Early Stopping)
    print(f"--- Optimal Epochs Found (Early Stopping): {best_epoch} ---")
    print(f"Minimum Error: {min_error:.4f}")
    print(f"Accuracy: {(1 - min_error):.4f}")
    print("Predictions at optimal epochs vs Target Values:")
    for i in range(len(best_predictions)):
        print(f"Predicted: {best_predictions[i][0]:.4f} - Target: {y[i][0]}")
    print("\n")

    # NEW: Print the absolute best results
    print(f"--- Absolute Best Results Overall ---")
    print(f"Achieved at Epoch: {absolute_best_epoch}")
    print(f"Lowest Error: {absolute_min_error:.4f}")
    print(f"Highest Accuracy: {absolute_max_accuracy:.4f}")
    print("\n")

    # plot error and accuracy as a function of epochs
    plt.figure(figsize=(10, 6))
    
    # Removed the point markers so the line is continuous and clean due to the large amount of data
    plt.plot(EPOCHS, error_values, color='red', linewidth=2, label='Error')
    plt.plot(EPOCHS, accuracy_values, color='blue', linewidth=2, label='Accuracy')
    
    # Add a vertical line marking exactly where the optimal epoch is
    plt.axvline(x=best_epoch, color='green', linestyle='--', label=f'Optimal: {best_epoch} Epochs')
    
    # NEW: Add a vertical dotted line marking the absolute best epoch
    plt.axvline(x=absolute_best_epoch, color='purple', linestyle=':', label=f'Absolute Best: {absolute_best_epoch} Epochs')
    
    plt.ylim(-0.05, 1.05)
    plt.title('Neural Network Performance: Error & Accuracy vs. Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Value [0 to 1]')
    plt.legend()
    plt.grid(True, which="both", ls="--", alpha=0.6)
    
    plt.tight_layout()
    plt.show()
```
