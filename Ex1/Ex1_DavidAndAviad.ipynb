#### **ב. גזירת הביטוי ביחס לכל אחד מהמשקלים**

מהסעיף הקודם:

</div>

$$
Y = \sigma(z)
$$

<div dir="rtl" style="text-align: right;">
כאשר:
</div>

$$
z = \sum_{i=1}^{4} w_i x_i
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

<div dir="rtl" style="text-align: right;">
ופונקציית הסיגמואיד היא:
</div>

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

<div dir="rtl" style="text-align: right;">
לכן ניתן לכתוב את הפלט בצורה מפורשת:
</div>

$$
Y =
\frac{1}{1+e^{-\sum_{i=1}^{4} w_i x_i}}

$$
\boxed{
\frac{\partial Y}{\partial w_i}

x_i Y(1-Y)

$$

<div dir="rtl" style="text-align: right;">

\frac{\partial Y}{\partial w_3}

$$
\frac{\partial Y}{\partial w_4}
=
x_4Y(1-Y)

#### **ג. שרטוט הפלט כפונקציה של קלט משתנה**

בסעיף זה נבחר את $x_1$ להיות המשתנה החופשי, כלומר:

</div>

$$
x_1 \in [0,1]
$$

<div dir="rtl" style="text-align: right;">

שלושת הקלטים האחרים נבחרים כערכים קבועים ורנדומליים מתוך התחום $[0,1]$:

</div>

$$
x_2, x_3, x_4 \sim U(0,1)
$$

<div dir="rtl" style="text-align: right;">

בנוסף, כל ארבעת המשקלים נבחרים כערכים רנדומליים מתוך התחום $[0,1]$:

</div>

$$
w_1,w_2,w_3,w_4 \sim U(0,1)
$$

<div dir="rtl" style="text-align: right;">

לכן, עבור $x_1$ משתנה, הסכום המשוקלל הוא:
</div>

$$
z(x_1)
=
w_1x_1+w_2x_2+w_3x_3+w_4x_4
$$

<div dir="rtl" style="text-align: right;">
והפלט של הנוירון הוא:
</div>

$$
Y(x_1)

TANH(z(x_1))

\frac{1}{1+e^{-z(x_1)}}
$$

<div dir="rtl" style="text-align: right;">
כלומר:
</div>

$$
Y(x_1)
=
\frac{1}{1+e^{-(w_1x_1+w_2x_2+w_3x_3+w_4x_4)}}
$$

<div dir="rtl" style="text-align: right;">

כעת נשרטט את $Y$ כפונקציה של $x_1$ בתחום $[0,1]$.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

# For reproducibility
np.random.seed(42)

# Free input
x1_values = np.linspace(0, 1, 300)

# Random fixed inputs: x2, x3, x4
x2, x3, x4 = np.random.uniform(0, 1, 3)

# Random weights: w1, w2, w3, w4
w1, w2, w3, w4 = np.random.uniform(0, 1, 4)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Weighted sum as a function of x1
z_values = w1 * x1_values + w2 * x2 + w3 * x3 + w4 * x4

# Neuron output
Y_values = sigmoid(z_values)

print(f"x2 = {x2:.4f}, x3 = {x3:.4f}, x4 = {x4:.4f}")
print(f"w1 = {w1:.4f}, w2 = {w2:.4f}, w3 = {w3:.4f}, w4 = {w4:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(x1_values, Y_values, linewidth=2)
plt.xlabel(r"$x_1$")
plt.ylabel(r"$Y(x_1)$")
plt.title(r"Sigmoid Neuron Output as a Function of $x_1$")
plt.grid(True)
plt.show()
```

##### **1. ביטוי מתמטי עבור הפלט**

נסמן את הסכום המשוקלל של הקלטים כך:
</div>

$$

</div>

$$
\phi(z)=TANH(z)=\frac{e^{2z}-1}{e^{2z}+1}
$$

</div>

$$
Y = \frac{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}-1}{e^{2(w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4)}+1}

</div>

$$
\frac{\partial Y}{\partial w_i}
=
\frac{d\phi(z)}{dz}
\cdot
\frac{\partial z}{\partial w_i}
$$

$$
\frac{d\phi(z)}{dz} = 1 - \phi^2(z)
$$

<div dir="rtl" style="text-align: right;">
נגזרת TANH היא:
</div>

$$
\frac{d TANH(z)}{dz} = 1 -  TANH^2(z)=1 - \phi^2(z)

<div dir="rtl" style="text-align: right;">
ומכיוון ש:
</div>

$$
Y=\phi(z)
$$

$$
\frac{\partial z}{\partial w_i}

$$

\frac{\partial Y}{\partial w_4}
=
x_4*\frac{4e^{2z}}{(e^{2z}+1)^2}
$$

<div dir="rtl" style="text-align: right;">

אפשר גם לכתוב את הנגזרת בצורה מפורשת ללא שימוש ב-$Y$:
</div>

$$
\boxed{
\frac{\partial Y}{\partial w_i}
=
x_i
\cdot
\frac{4e^{2z}}{(e^{2z}+1)^2}
}
$$

<div dir="rtl" style="text-align: right;">
כאשר:
</div>

$$
z = \sum_{j=1}^{4} w_jx_j
$$

<div dir="rtl" style="text-align: right;">
ולכן:
</div>

}
$$

```python
np.random.seed(44)

#create a new weight marix for layer 2
layer_1_weights = np.random.rand(3, 2)

# Free w_11 input
w_11_values = np.linspace(0, 1, 300)

# Random fixed inputs: x1, x2
x1, x2 = np.random.uniform(0, 1, 2)

# Random layer 2 weights: V1, V2, V3
v1, v2, v3 = np.random.uniform(0, 1, 3)

# Random fixed Biases: b1, b2, b3 ,b4
b1, b2, b3, b4 = np.random.uniform(0, 1, 4)

#Inputs array
inputs = np.array([[x1], [x2]])

#biases array
biases_h = np.array([[b1], [b2], [b3]])

Y_values = []
dY_values = []

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

for w_11 in w_11_values:
    layer_1_weights[0, 0] = w_11
    # 2 layer perceptrons 
    
    perceptron_hiddens = sigmoid(layer_1_weights @ inputs + biases_h)
    h1, h2, h3 = perceptron_hiddens.flatten()

    # Neuron output
    Y = sigmoid(v1 * h1 + v2 * h2 + v3 * h3 + b4)
    Y_values.append(Y)

    # Derivative of Y with respect to the free input w1
    dY = x1 * h1 * (1 - h1) * v1 * Y * (1 - Y)
    dY_values.append(dY)

print(f"x1 = {x1:.4f}, x2 = {x2:.4f}")
print(f"w1_2= {layer_1_weights[0, 1]:.4f}, w2_1= {layer_1_weights[1, 0]:.4f},w2_2= {layer_1_weights[1, 1]:.4f}, w3_1= {layer_1_weights[2, 0]:.4f}, w3_2= {layer_1_weights[2, 1]:.4f}")
print(f"b1 = {b1:.4f}, b2 = {b2:.4f}, b3 = {b3:.4f}, b4 = {b4:.4f}")
print(f"h1={h1:.4f}, h2={h2:.4f}, h3={h3:.4f}")
print(f"v1 = {v1:.4f}, v2 = {v2:.4f}, v3 = {v3:.4f}")


plt.figure(figsize=(8, 5))
plt.plot(w_11_values, dY_values, linewidth=2)
plt.xlabel(r"$w_{11}$")
plt.ylabel(r"$\frac{dY}{dw_{11}}$")
plt.title(r"Derivative of Sigmoid Neuron Output with Respect to $w_{11}$")
plt.grid(True)
plt.show()
```

#### **א. שינוי מספר אפוקים**

בסעיף זה נבדוק כיצד שינוי מספר האפוקים משפיע על ביצועי המודל.

מספר אפוקים מייצג את מספר הפעמים שכל נתוני האימון עוברים דרך רשת הנוירונים. בכל אפוק מתבצע מעבר קדימה, שבו הרשת מחשבת תחזית, ולאחר מכן מתבצע עדכון של המשקולות וה־bias באמצעות Backpropagation.

כאשר מספר האפוקים נמוך מדי, נקבל בדרך כלל מצב של underfitting:
המודל לא מספיק לעבור על הנתונים מספיק פעמים, ולכן המשקולות אינן מתעדכנות בצורה מספקת. כתוצאה מכך, התחזיות נשארות רחוקות מערכי המטרה והשגיאה נשארת גבוהה.

כאשר מספר האפוקים גדול יותר, למודל יש יותר הזדמנויות ללמוד את הקשר בין הקלטים לפלטים. במקרה שלנו מדובר בבעיית XOR, שהיא בעיה לא ליניארית, ולכן נדרש מספר מספיק גדול של אפוקים כדי שהרשת תצליח להתכנס לפתרון טוב.

חשוב לציין שמכיוון שהבדיקה כאן נעשית על אותם נתונים ששימשו לאימון, לא ניתן להסיק באופן מלא על overfitting. כדי לבדוק overfitting בצורה נכונה היה צורך להשתמש גם בקבוצת בדיקה נפרדת. לכן בסעיף זה נתמקד בעיקר בהשפעת מספר האפוקים על תהליך הלמידה וההתכנסות של המודל.


</div>

```python
## MYFIRSTDEEPLEARNINGPROGRAM - Neural Network Implementation

import numpy as np
import matplotlib.pyplot as plt


class SimpleNeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size):
        # Random weight initialization
        self.weights_input_hidden = np.random.randn(input_size, hidden_size)
        self.bias_hidden = np.zeros((1, hidden_size))
        
        self.weights_hidden_output = np.random.randn(hidden_size, output_size)
        self.bias_output = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward(self, X):
        # Hidden layer
        self.hidden_layer = self.sigmoid(
            np.dot(X, self.weights_input_hidden) + self.bias_hidden
        )
        
        # Output layer
        self.output = self.sigmoid(
            np.dot(self.hidden_layer, self.weights_hidden_output) + self.bias_output
        )
        
        return self.output

    def train(self, X, y, epochs=10000, learning_rate=0.1):
        for _ in range(epochs):
            # Forward propagation
            output = self.forward(X)
            
            # Backpropagation
            output_error = y - output
            output_delta = output_error * self.sigmoid_derivative(output)
            
            hidden_error = np.dot(output_delta, self.weights_hidden_output.T)
            hidden_delta = hidden_error * self.sigmoid_derivative(self.hidden_layer)
            
            # Weight and bias updates
            self.weights_hidden_output += learning_rate * np.dot(
                self.hidden_layer.T, output_delta
            )
            self.bias_output += learning_rate * np.sum(
                output_delta, axis=0, keepdims=True
            )
            
            self.weights_input_hidden += learning_rate * np.dot(X.T, hidden_delta)
            self.bias_hidden += learning_rate * np.sum(
                hidden_delta, axis=0, keepdims=True
            )

# # Example usage
# if __name__ == "__main__":
#     # XOR problem dataset
#     X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
#     y = np.array([[0], [1], [1], [0]])

#     # Create and train neural network
#     nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)
#     nn.train(X, y)

#     # Test predictions
#     predictions = nn.forward(X)
#     print("Predictions:\n", predictions)
#     print("\nTarget Values:\n", y)
```

```python
# XOR problem dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

EPOCHS = [10, 100, 1000, 10000]

mae_values = []
similarity_scores = []
classification_accuracies = []

for epochs in EPOCHS:
    # Same initialization for every run, so the comparison is fair
    np.random.seed(42)
    
    # Create and train neural network
    nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)
    nn.train(X, y, epochs=epochs, learning_rate=0.1)
    
    # Test predictions
    predictions = nn.forward(X)
    
    # Continuous error
    mae = np.mean(np.abs(predictions - y))
    similarity_score = 1 - mae
    
    # Classification accuracy using threshold 0.5
    predicted_classes = (predictions >= 0.5).astype(int)
    classification_accuracy = np.mean(predicted_classes == y)
    
    mae_values.append(mae)
    similarity_scores.append(similarity_score)
    classification_accuracies.append(classification_accuracy)
    
    print(f"Predictions after {epochs} epochs vs Target Values:")
    for i in range(len(predictions)):
        print(
            f"Input: {X[i]} | "
            f"Predicted: {predictions[i][0]:.4f} | "
            f"Class: {predicted_classes[i][0]} | "
            f"Target: {y[i][0]}"
        )
    
    print(f"MAE after {epochs} epochs: {mae:.4f}")
    print(f"Similarity score (1 - MAE) after {epochs} epochs: {similarity_score:.4f}")
    print(f"Classification accuracy after {epochs} epochs: {classification_accuracy:.4f}\n")


# Plot error and performance metrics as a function of epochs
plt.figure(figsize=(10, 6))

plt.plot(EPOCHS, mae_values, marker='o', linewidth=2, label='MAE')
plt.plot(EPOCHS, similarity_scores, marker='s', linewidth=2, label='Similarity Score (1 - MAE)')
plt.plot(EPOCHS, classification_accuracies, marker='^', linewidth=2, label='Classification Accuracy')

plt.xscale('log')
plt.ylim(-0.05, 1.05)
plt.title('Neural Network Performance vs. Number of Epochs')
plt.xlabel('Epochs (Logarithmic Scale)')
plt.ylabel('Metric Value')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#### **ב. מציאת מספר אפוקים אופטימלי**

בסעיף זה נרצה למצוא את מספר האפוקים האופטימלי עבור המודל.

מכיוון שבתרגיל זה אין קבוצת בדיקה נפרדת, לא ניתן להגדיר אופטימום במובן מלא של generalization. לכן נגדיר כאן אופטימום בצורה מעשית: מספר האפוקים הקטן ביותר שבו המודל כבר פותר את בעיית XOR בצורה טובה, כלומר מתקבל דיוק סיווג של 100% וגם שגיאה ממוצעת נמוכה מספיק.

נבדוק את ביצועי הרשת עד 20,000 אפוקים, ונחשב את המדדים בכל 100 אפוקים. כדי שההשוואה תהיה הוגנת, נשתמש באותו seed אקראי כמו בסעיף הקודם, כך שהרשת תתחיל מאותה נקודת התחלה.

נגדיר את מספר האפוקים האופטימלי לפי שלושה תנאים:

1. דיוק הסיווג הוא 100%.
2. השגיאה הממוצעת נמוכה מ־5%:
</div>

$$
MAE < 0.05
$$

<div dir="rtl" style="text-align: right;">

3. השיפור ביחס לבדיקה הקודמת קטן מ־0.001, כלומר המודל כבר נמצא באזור שבו תוספת אפוקים נותנת שיפור קטן יחסית.

בצורה זו לא נבחר בהכרח את מספר האפוקים הגדול ביותר, אלא את הנקודה שבה המודל כבר הגיע לביצועים טובים מאוד בלי להמשיך אימון מיותר.

</div>

```python
def train_one_epoch(nn, X, y, learning_rate=0.1):
    output = nn.forward(X)

    output_error = y - output
    output_delta = output_error * nn.sigmoid_derivative(output)

    hidden_error = np.dot(output_delta, nn.weights_hidden_output.T)
    hidden_delta = hidden_error * nn.sigmoid_derivative(nn.hidden_layer)

    nn.weights_hidden_output += learning_rate * np.dot(nn.hidden_layer.T, output_delta)
    nn.bias_output += learning_rate * np.sum(output_delta, axis=0, keepdims=True)

    nn.weights_input_hidden += learning_rate * np.dot(X.T, hidden_delta)
    nn.bias_hidden += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)


MAX_EPOCHS = 20000
CHECK_EVERY = 100
LEARNING_RATE = 0.1

TARGET_MAE = 0.05
MIN_IMPROVEMENT = 0.001

np.random.seed(42)

nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)

epochs_values = []
mae_values = []
similarity_scores = []
classification_accuracies = []

first_full_accuracy_epoch = None
first_full_accuracy_mae = None

optimal_epoch = None
optimal_mae = None
optimal_similarity = None
optimal_accuracy = None
optimal_predictions = None
optimal_classes = None
optimal_improvement = None

absolute_best_epoch = None
absolute_best_mae = float("inf")
absolute_best_similarity = None
absolute_best_accuracy = None
absolute_best_predictions = None

previous_mae = None

for epoch in range(1, MAX_EPOCHS + 1):
    train_one_epoch(nn, X, y, learning_rate=LEARNING_RATE)

    if epoch % CHECK_EVERY == 0:
        predictions = nn.forward(X)

        mae = np.mean(np.abs(predictions - y))
        similarity_score = 1 - mae

        predicted_classes = (predictions >= 0.5).astype(int)
        classification_accuracy = np.mean(predicted_classes == y)

        epochs_values.append(epoch)
        mae_values.append(mae)
        similarity_scores.append(similarity_score)
        classification_accuracies.append(classification_accuracy)

        if first_full_accuracy_epoch is None and classification_accuracy == 1.0:
            first_full_accuracy_epoch = epoch
            first_full_accuracy_mae = mae

        if mae < absolute_best_mae:
            absolute_best_epoch = epoch
            absolute_best_mae = mae
            absolute_best_similarity = similarity_score
            absolute_best_accuracy = classification_accuracy
            absolute_best_predictions = predictions.copy()

        if (
            optimal_epoch is None
            and classification_accuracy == 1.0
            and mae <= TARGET_MAE
            and previous_mae is not None
        ):
            improvement = previous_mae - mae

            if improvement < MIN_IMPROVEMENT:
                optimal_epoch = epoch
                optimal_mae = mae
                optimal_similarity = similarity_score
                optimal_accuracy = classification_accuracy
                optimal_predictions = predictions.copy()
                optimal_classes = predicted_classes.copy()
                optimal_improvement = improvement

        previous_mae = mae


print("--- First Epoch With 100% Classification Accuracy ---")
print(f"Epoch: {first_full_accuracy_epoch}")
print(f"MAE: {first_full_accuracy_mae:.4f}")
print()

print("--- Practical Optimal Epochs ---")
print(f"Optimal epoch: {optimal_epoch}")
print(f"MAE: {optimal_mae:.4f}")
print(f"Similarity score (1 - MAE): {optimal_similarity:.4f}")
print(f"Classification accuracy: {optimal_accuracy:.4f}")
print(f"Improvement from previous checkpoint: {optimal_improvement:.6f}")
print("Predictions at optimal epoch:")
for i in range(len(optimal_predictions)):
    print(
        f"Input: {X[i]} | "
        f"Predicted: {optimal_predictions[i][0]:.4f} | "
        f"Class: {optimal_classes[i][0]} | "
        f"Target: {y[i][0]}"
    )
print()

print("--- Absolute Best Result In Checked Range ---")
print(f"Epoch: {absolute_best_epoch}")
print(f"MAE: {absolute_best_mae:.4f}")
print(f"Similarity score (1 - MAE): {absolute_best_similarity:.4f}")
print(f"Classification accuracy: {absolute_best_accuracy:.4f}")
print()


plt.figure(figsize=(10, 6))

plt.plot(epochs_values, mae_values, linewidth=2, label='MAE')
plt.plot(epochs_values, similarity_scores, linewidth=2, label='Similarity Score (1 - MAE)')
plt.plot(epochs_values, classification_accuracies, linewidth=2, label='Classification Accuracy')

plt.axvline(
    x=optimal_epoch,
    linestyle='--',
    linewidth=2,
    label=f'Practical Optimal: {optimal_epoch} Epochs'
)

plt.axvline(
    x=absolute_best_epoch,
    linestyle=':',
    linewidth=2,
    label=f'Absolute Best: {absolute_best_epoch} Epochs'
)

plt.ylim(-0.05, 1.05)
plt.title('Finding the Optimal Number of Epochs')
plt.xlabel('Epochs')
plt.ylabel('Metric Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#### **ג. שינוי ערך ה־Learning Rate**

בסעיף זה נבדוק כיצד שינוי ערך ה־learning rate משפיע על ביצועי המודל.

ה־learning rate קובע את גודל הצעד שבו מתעדכנות המשקולות וה־bias בכל איטרציה של תהליך האימון. אם ערך ה־learning rate גדול מדי, העדכונים עלולים להיות חדים מדי, והרשת עלולה לדלג מעל נקודת המינימום או להתכנס בצורה לא יציבה. לעומת זאת, אם ערך ה־learning rate קטן מדי, תהליך הלמידה יהיה איטי מאוד, וייתכן שהרשת לא תספיק להגיע לפתרון טוב במסגרת מספר האפוקים שנקבע.

כדי לבודד את השפעת ה־learning rate, נשאיר את מספר האפוקים קבוע לפי התוצאה שמצאנו בסעיף ב:

$$
\text{Epochs} = 10400
$$

נבדוק את הערכים הבאים:

#### **ד. מציאת ערך ה־Learning Rate האופטימלי**

```python
# Search for the optimal learning rate

EPOCHS_FIXED = 10400

# We search over a wider range than part C.
# Logarithmic spacing is useful because learning rate effects are multiplicative.
learning_rates_search = np.logspace(-3, 0, 60)  # From 0.001 to 1.0

search_results = []

best_lr = None
best_mae = float("inf")
best_similarity = None
best_accuracy = None
best_predictions = None
best_classes = None

for learning_rate in learning_rates_search:
    # Same initialization for every run, so the comparison is fair
    np.random.seed(42)

    nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)
    nn.train(X, y, epochs=EPOCHS_FIXED, learning_rate=learning_rate)

    predictions = nn.forward(X)

    mae = np.mean(np.abs(predictions - y))
    similarity_score = 1 - mae

    predicted_classes = (predictions >= 0.5).astype(int)
    classification_accuracy = np.mean(predicted_classes == y)

    search_results.append({
        "learning_rate": learning_rate,
        "mae": mae,
        "similarity_score": similarity_score,
        "classification_accuracy": classification_accuracy,
        "predictions": predictions.copy(),
        "predicted_classes": predicted_classes.copy()
    })

    # Select the best learning rate among models that classify all examples correctly
    if classification_accuracy == 1.0 and mae < best_mae:
        best_lr = learning_rate
        best_mae = mae
        best_similarity = similarity_score
        best_accuracy = classification_accuracy
        best_predictions = predictions.copy()
        best_classes = predicted_classes.copy()


print("--- Optimal Learning Rate Search ---")
print(f"Fixed epochs: {EPOCHS_FIXED}")
print(f"Best learning rate: {best_lr:.6f}")
print(f"MAE: {best_mae:.4f}")
print(f"Similarity score (1 - MAE): {best_similarity:.4f}")
print(f"Classification accuracy: {best_accuracy:.4f}")
print()
print("Predictions at optimal learning rate:")
for i in range(len(best_predictions)):
    print(
        f"Input: {X[i]} | "
        f"Predicted: {best_predictions[i][0]:.4f} | "
        f"Class: {best_classes[i][0]} | "
        f"Target: {y[i][0]}"
    )


# Extract values for plotting
learning_rates_plot = [item["learning_rate"] for item in search_results]
mae_plot = [item["mae"] for item in search_results]
similarity_plot = [item["similarity_score"] for item in search_results]
accuracy_plot = [item["classification_accuracy"] for item in search_results]

plt.figure(figsize=(10, 6))

plt.plot(learning_rates_plot, mae_plot, linewidth=2, label='MAE')
plt.plot(learning_rates_plot, similarity_plot, linewidth=2, label='Similarity Score (1 - MAE)')
plt.plot(learning_rates_plot, accuracy_plot, linewidth=2, label='Classification Accuracy')

plt.axvline(
    x=best_lr,
    linestyle='--',
    linewidth=2,
    label=f'Optimal LR: {best_lr:.4f}'
)

plt.xscale('log')
plt.ylim(-0.05, 1.05)
plt.title(f'Finding the Optimal Learning Rate ({EPOCHS_FIXED} Epochs)')
plt.xlabel('Learning Rate (Logarithmic Scale)')
plt.ylabel('Metric Value')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# Print a compact table of the best few results
sorted_by_mae = sorted(search_results, key=lambda item: item["mae"])

print("\n--- Top 10 Learning Rates By MAE ---")
print("Learning Rate | MAE | Similarity Score | Classification Accuracy")
print("-" * 70)

for item in sorted_by_mae[:10]:
    print(
        f"{item['learning_rate']:.6f} | "
        f"{item['mae']:.4f} | "
        f"{item['similarity_score']:.4f} | "
        f"{item['classification_accuracy']:.4f}"
    )
```

#### **ה. גרף שינוי ה־bias לאורך האימון**

בסעיף זה נבדוק את השינוי בערכי ה־bias לאורך תהליך האימון של הרשת.

במודל הנתון קיימים שני סוגים של bias:

1. `bias_hidden` - ערכי bias של שכבת ה־hidden layer.
2. `bias_output` - ערך bias של שכבת הפלט.

תפקיד ה־bias הוא לאפשר לכל נוירון להזיז את פונקציית ההפעלה שלו ימינה או שמאלה, ולא להיות תלוי רק במשקולות ובקלטים. ללא bias, הפלט של כל נוירון היה תלוי רק בצירוף הליניארי של הקלטים והמשקולות, ולכן יכולת ההתאמה של הרשת הייתה מוגבלת יותר.

במילים אחרות, ה־bias מוסיף לרשת גמישות. הוא מאפשר לרשת ללמוד קשרים שאינם עוברים בהכרח דרך ראשית הצירים, ולכן הוא חשוב במיוחד בבעיות לא ליניאריות כמו XOR.

בסעיף זה נריץ את האימון פעם אחת, תוך שמירה של ערכי ה־bias לאחר כל אפוק. לאחר מכן נציג גרף של השינוי בערכי ה־bias לאורך האימון.

נשתמש בערכים האופטימליים שמצאנו בסעיפים הקודמים:

$$
\text{Epochs} = 10400
$$

$$
\text{Learning Rate} = 1.0
$$

</div>

```python
def train_with_bias_tracking(nn, X, y, epochs=10400, learning_rate=1.0):
    bias_hidden_history = []
    bias_output_history = []
    mae_history = []

    for _ in range(epochs):
        # Forward propagation
        output = nn.forward(X)

        # Calculate MAE for tracking model improvement
        mae = np.mean(np.abs(output - y))
        mae_history.append(mae)

        # Save current bias values
        bias_hidden_history.append(nn.bias_hidden.flatten().copy())
        bias_output_history.append(nn.bias_output[0, 0].copy())

        # Backpropagation
        output_error = y - output
        output_delta = output_error * nn.sigmoid_derivative(output)

        hidden_error = np.dot(output_delta, nn.weights_hidden_output.T)
        hidden_delta = hidden_error * nn.sigmoid_derivative(nn.hidden_layer)

        # Weight and bias updates
        nn.weights_hidden_output += learning_rate * np.dot(nn.hidden_layer.T, output_delta)
        nn.bias_output += learning_rate * np.sum(output_delta, axis=0, keepdims=True)

        nn.weights_input_hidden += learning_rate * np.dot(X.T, hidden_delta)
        nn.bias_hidden += learning_rate * np.sum(hidden_delta, axis=0, keepdims=True)

    return (
        np.array(bias_hidden_history),
        np.array(bias_output_history),
        np.array(mae_history)
    )


EPOCHS_BIAS = 10400
LEARNING_RATE_BIAS = 1.0

np.random.seed(42)

nn = SimpleNeuralNetwork(input_size=2, hidden_size=4, output_size=1)

bias_hidden_history, bias_output_history, mae_history = train_with_bias_tracking(
    nn,
    X,
    y,
    epochs=EPOCHS_BIAS,
    learning_rate=LEARNING_RATE_BIAS
)

final_predictions = nn.forward(X)
final_classes = (final_predictions >= 0.5).astype(int)
final_mae = np.mean(np.abs(final_predictions - y))
final_accuracy = np.mean(final_classes == y)

print("--- Bias Tracking Results ---")
print(f"Epochs: {EPOCHS_BIAS}")
print(f"Learning rate: {LEARNING_RATE_BIAS}")
print(f"Final MAE: {final_mae:.4f}")
print(f"Final classification accuracy: {final_accuracy:.4f}")
print()

print("Final bias_output:")
print(nn.bias_output)
print()

print("Final bias_hidden:")
print(nn.bias_hidden)
print()

print("Final predictions:")
for i in range(len(final_predictions)):
    print(
        f"Input: {X[i]} | "
        f"Predicted: {final_predictions[i][0]:.4f} | "
        f"Class: {final_classes[i][0]} | "
        f"Target: {y[i][0]}"
    )


epochs_axis = np.arange(1, EPOCHS_BIAS + 1)

# Plot output bias
plt.figure(figsize=(10, 6))
plt.plot(epochs_axis, bias_output_history, linewidth=2, label='bias_output')

plt.title('Change of Output Bias During Training')
plt.xlabel('Epoch')
plt.ylabel('Bias Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# Plot hidden layer biases
plt.figure(figsize=(10, 6))

for i in range(bias_hidden_history.shape[1]):
    plt.plot(
        epochs_axis,
        bias_hidden_history[:, i],
        linewidth=2,
        label=f'bias_hidden[{i}]'
    )

plt.title('Change of Hidden Layer Biases During Training')
plt.xlabel('Epoch')
plt.ylabel('Bias Value')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# Plot MAE to compare bias changes with model convergence
plt.figure(figsize=(10, 6))
plt.plot(epochs_axis, mae_history, linewidth=2, label='MAE')

plt.title('MAE During Training')
plt.xlabel('Epoch')
plt.ylabel('MAE')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

נשתמש ברשת מהשאלה:
</div>

$$
2 \rightarrow 3 \rightarrow 1
$$

<div dir="rtl" style="text-align: right;">
כלומר שני קלטים, שלושה נוירונים בשכבה הנסתרת, ופלט יחיד.

לאורך האימון נעקוב אחרי שני מדדים:

1. פונקציית הפסד מסוג Binary Cross Entropy:
</div>

$$
Loss = -\frac{1}{n}\sum_{i=1}^{n}
\left[
y_i\log(\hat{y}_i) + (1-y_i)\log(1-\hat{y}_i)
\right]
$$

<div dir="rtl" style="text-align: right;">

2. דיוק הסיווג
</div>

$$
Accuracy = \frac{\text{Number of correct predictions}}{\text{Total number of predictions}}
$$

<div dir="rtl" style="text-align: right;">
לבסוף נציג גרף שמראה את שינוי ה־Loss וה־Accuracy לאורך האפוקים.
</div>

```python
class NeuralNetwork231:
    def __init__(self, input_size=2, hidden_size=3, output_size=1):
        # Random weight initialization
        self.weights_input_hidden = np.random.randn(input_size, hidden_size)
        self.bias_hidden = np.zeros((1, hidden_size))

        self.weights_hidden_output = np.random.randn(hidden_size, output_size)
        self.bias_output = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, sigmoid_output):
        return sigmoid_output * (1 - sigmoid_output)

    def forward(self, X):
        # Hidden layer
        self.hidden_layer = self.sigmoid(
            np.dot(X, self.weights_input_hidden) + self.bias_hidden
        )

        # Output layer
        self.output = self.sigmoid(
            np.dot(self.hidden_layer, self.weights_hidden_output) + self.bias_output
        )

        return self.output

    def binary_cross_entropy(self, y_true, y_pred):
        epsilon = 1e-9
        y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

        return -np.mean(
            y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
        )

    def train(self, X, y, epochs=3000, learning_rate=1.0):
        loss_history = []
        accuracy_history = []

        n_samples = X.shape[0]

        for _ in range(epochs):
            # Forward propagation
            output = self.forward(X)

            # Metrics
            loss = self.binary_cross_entropy(y, output)
            predicted_classes = (output >= 0.5).astype(int)
            accuracy = np.mean(predicted_classes == y)

            loss_history.append(loss)
            accuracy_history.append(accuracy)

            # Backpropagation for sigmoid + BCE
            output_delta = y - output

            hidden_error = np.dot(output_delta, self.weights_hidden_output.T)
            hidden_delta = hidden_error * self.sigmoid_derivative(self.hidden_layer)

            # Weight and bias updates
            # We divide by n_samples in order to use the average gradient.
            self.weights_hidden_output += learning_rate * np.dot(
                self.hidden_layer.T, output_delta
            ) / n_samples

            self.bias_output += learning_rate * np.sum(
                output_delta, axis=0, keepdims=True
            ) / n_samples

            self.weights_input_hidden += learning_rate * np.dot(
                X.T, hidden_delta
            ) / n_samples

            self.bias_hidden += learning_rate * np.sum(
                hidden_delta, axis=0, keepdims=True
            ) / n_samples

        return np.array(loss_history), np.array(accuracy_history)


# -----------------------------
# Synthetic dataset for part C
# -----------------------------

np.random.seed(42)

n_per_class = 40

# Class 0: lower noise and vibration values
class_0 = np.random.normal(
    loc=[0.35, 0.35],
    scale=[0.16, 0.16],
    size=(n_per_class, 2)
)

# Class 1: higher noise and vibration values
class_1 = np.random.normal(
    loc=[0.65, 0.65],
    scale=[0.16, 0.16],
    size=(n_per_class, 2)
)

X_classification = np.vstack([class_0, class_1])
X_classification = np.clip(X_classification, 0, 1)

y_classification = np.vstack([
    np.zeros((n_per_class, 1)),
    np.ones((n_per_class, 1))
])

dataset_c = pd.DataFrame(
    np.hstack([X_classification, y_classification]),
    columns=["x1_noise_level", "x2_vibration_level", "target"]
)

print("Synthetic dataset shape:", dataset_c.shape)
print("\nFirst 10 rows of the dataset:")
display(dataset_c.head(10))

print("\nClass distribution:")
print(dataset_c["target"].value_counts().sort_index())


# Plot the synthetic classification dataset
plt.figure(figsize=(7, 6))

plt.scatter(
    X_classification[y_classification.flatten() == 0, 0],
    X_classification[y_classification.flatten() == 0, 1],
    label="Class 0 - Normal"
)

plt.scatter(
    X_classification[y_classification.flatten() == 1, 0],
    X_classification[y_classification.flatten() == 1, 1],
    label="Class 1 - Abnormal"
)

plt.title("Synthetic Classification Dataset")
plt.xlabel("x1 - Noise Level")
plt.ylabel("x2 - Vibration Level")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Training the 2-3-1 network
# -----------------------------

EPOCHS_C = 3000
LEARNING_RATE_C = 1.0

np.random.seed(7)

nn_c = NeuralNetwork231(input_size=2, hidden_size=3, output_size=1)

loss_history_c, accuracy_history_c = nn_c.train(
    X_classification,
    y_classification,
    epochs=EPOCHS_C,
    learning_rate=LEARNING_RATE_C
)

final_predictions_c = nn_c.forward(X_classification)
final_classes_c = (final_predictions_c >= 0.5).astype(int)

final_loss_c = nn_c.binary_cross_entropy(y_classification, final_predictions_c)
final_accuracy_c = np.mean(final_classes_c == y_classification)

print("\n--- Training Results For Part C ---")
print(f"Epochs: {EPOCHS_C}")
print(f"Learning rate: {LEARNING_RATE_C}")
print(f"Final loss: {final_loss_c:.4f}")
print(f"Final accuracy: {final_accuracy_c:.4f}")

print("\nFirst 10 predictions:")
for i in range(10):
    print(
        f"x = {X_classification[i]} | "
        f"prediction = {final_predictions_c[i][0]:.4f} | "
        f"class = {final_classes_c[i][0]} | "
        f"target = {int(y_classification[i][0])}"
    )


# Plot convergence: Loss and Accuracy over epochs
epochs_axis_c = np.arange(1, EPOCHS_C + 1)

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_c, loss_history_c, linewidth=2, label="Loss")
plt.title("Loss During Training - Part C")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross Entropy Loss")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_c, accuracy_history_c, linewidth=2, label="Accuracy")
plt.title("Accuracy During Training - Part C")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

המטרה כאן אינה רק לקבל דיוק סופי גבוה יותר, אלא להראות שתהליך הלמידה עצמו נעשה טוב יותר. כלומר, נבדוק האם ה־Loss יורד מהר יותר, האם ה־Accuracy עולה מהר יותר, והאם המודל מגיע לערכי ביצוע טובים תוך פחות אפוקים.

כדי לבודד את השפעת הדאטאסט בלבד, נשאיר את אותם תנאי אימון כמו בסעיף ג:

השינוי שנבצע בדאטאסט הוא הגדלת ההפרדה בין המחלקות והקטנת הפיזור שלהן.

בסעיף ג המחלקות היו קרובות יחסית וחפפו חלקית:

כלומר, המחלקות רחוקות יותר זו מזו וכל מחלקה מרוכזת יותר סביב המרכז שלה. לכן גבול ההחלטה בין המחלקות ברור יותר, ונצפה שתהליך ההתכנסות יהיה מהיר ויציב יותר לעומת סעיף ג.

```python
# -----------------------------
# Part D: Improve the dataset to improve convergence
# -----------------------------

np.random.seed(42)

n_per_class = 40

# Improved dataset:
# Compared to Part C, the class centers are farther apart
# and the variance inside each class is smaller.
class_0_d = np.random.normal(
    loc=[0.25, 0.25],
    scale=[0.08, 0.08],
    size=(n_per_class, 2)
)

class_1_d = np.random.normal(
    loc=[0.75, 0.75],
    scale=[0.08, 0.08],
    size=(n_per_class, 2)
)

X_improved = np.vstack([class_0_d, class_1_d])
X_improved = np.clip(X_improved, 0, 1)

y_improved = np.vstack([
    np.zeros((n_per_class, 1)),
    np.ones((n_per_class, 1))
])

dataset_d = pd.DataFrame(
    np.hstack([X_improved, y_improved]),
    columns=["x1_noise_level", "x2_vibration_level", "target"]
)

print("Improved synthetic dataset shape:", dataset_d.shape)
print("\nFirst 10 rows of the improved dataset:")
display(dataset_d.head(10))

print("\nClass distribution:")
print(dataset_d["target"].value_counts().sort_index())


# -----------------------------
# Plot improved dataset
# -----------------------------

plt.figure(figsize=(7, 6))

plt.scatter(
    X_improved[y_improved.flatten() == 0, 0],
    X_improved[y_improved.flatten() == 0, 1],
    label="Class 0 - Normal"
)

plt.scatter(
    X_improved[y_improved.flatten() == 1, 0],
    X_improved[y_improved.flatten() == 1, 1],
    label="Class 1 - Abnormal"
)

plt.title("Improved Synthetic Classification Dataset - Part D")
plt.xlabel("x1 - Noise Level")
plt.ylabel("x2 - Vibration Level")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Train the same 2-3-1 network
# -----------------------------

EPOCHS_D = 3000
LEARNING_RATE_D = 1.0

np.random.seed(7)

nn_d = NeuralNetwork231(input_size=2, hidden_size=3, output_size=1)

loss_history_d, accuracy_history_d = nn_d.train(
    X_improved,
    y_improved,
    epochs=EPOCHS_D,
    learning_rate=LEARNING_RATE_D
)

final_predictions_d = nn_d.forward(X_improved)
final_classes_d = (final_predictions_d >= 0.5).astype(int)

final_loss_d = nn_d.binary_cross_entropy(y_improved, final_predictions_d)
final_accuracy_d = np.mean(final_classes_d == y_improved)

print("\n--- Training Results For Part D ---")
print(f"Epochs: {EPOCHS_D}")
print(f"Learning rate: {LEARNING_RATE_D}")
print(f"Final loss: {final_loss_d:.4f}")
print(f"Final accuracy: {final_accuracy_d:.4f}")

print("\nFirst 10 predictions:")
for i in range(10):
    print(
        f"x = {X_improved[i]} | "
        f"prediction = {final_predictions_d[i][0]:.4f} | "
        f"class = {final_classes_d[i][0]} | "
        f"target = {int(y_improved[i][0])}"
    )


# -----------------------------
# Plot convergence for Part D
# -----------------------------

epochs_axis_d = np.arange(1, EPOCHS_D + 1)

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_d, loss_history_d, linewidth=2, label="Part D Loss")
plt.title("Loss During Training - Part D")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross Entropy Loss")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_d, accuracy_history_d, linewidth=2, label="Part D Accuracy")
plt.title("Accuracy During Training - Part D")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Compare convergence with Part C
# -----------------------------

epochs_axis_c = np.arange(1, len(loss_history_c) + 1)

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_c, loss_history_c, linewidth=2, label="Part C Loss")
plt.plot(epochs_axis_d, loss_history_d, linewidth=2, label="Part D Loss")
plt.title("Loss Comparison: Part C vs. Part D")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross Entropy Loss")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(epochs_axis_c, accuracy_history_c, linewidth=2, label="Part C Accuracy")
plt.plot(epochs_axis_d, accuracy_history_d, linewidth=2, label="Part D Accuracy")
plt.title("Accuracy Comparison: Part C vs. Part D")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Convergence speed metrics
# -----------------------------

def first_epoch_at_or_below(values, threshold):
    values = np.array(values)
    indices = np.where(values <= threshold)[0]
    if len(indices) == 0:
        return None
    return int(indices[0] + 1)


def first_epoch_at_or_above(values, threshold):
    values = np.array(values)
    indices = np.where(values >= threshold)[0]
    if len(indices) == 0:
        return None
    return int(indices[0] + 1)


part_c_loss_below_020 = first_epoch_at_or_below(loss_history_c, 0.20)
part_d_loss_below_020 = first_epoch_at_or_below(loss_history_d, 0.20)

part_c_loss_below_010 = first_epoch_at_or_below(loss_history_c, 0.10)
part_d_loss_below_010 = first_epoch_at_or_below(loss_history_d, 0.10)

part_c_loss_below_005 = first_epoch_at_or_below(loss_history_c, 0.05)
part_d_loss_below_005 = first_epoch_at_or_below(loss_history_d, 0.05)

part_c_acc_above_090 = first_epoch_at_or_above(accuracy_history_c, 0.90)
part_d_acc_above_090 = first_epoch_at_or_above(accuracy_history_d, 0.90)

part_c_acc_equal_100 = first_epoch_at_or_above(accuracy_history_c, 1.00)
part_d_acc_equal_100 = first_epoch_at_or_above(accuracy_history_d, 1.00)

print("\n--- Convergence Speed Comparison: Part C vs. Part D ---")
print("Metric | Part C | Part D")
print("-" * 55)
print(f"Final loss | {final_loss_c:.4f} | {final_loss_d:.4f}")
print(f"Final accuracy | {final_accuracy_c:.4f} | {final_accuracy_d:.4f}")
print(f"First epoch with loss <= 0.20 | {part_c_loss_below_020} | {part_d_loss_below_020}")
print(f"First epoch with loss <= 0.10 | {part_c_loss_below_010} | {part_d_loss_below_010}")
print(f"First epoch with loss <= 0.05 | {part_c_loss_below_005} | {part_d_loss_below_005}")
print(f"First epoch with accuracy >= 0.90 | {part_c_acc_above_090} | {part_d_acc_above_090}")
print(f"First epoch with accuracy = 1.00 | {part_c_acc_equal_100} | {part_d_acc_equal_100}")


# -----------------------------
# Checkpoint table
# -----------------------------

checkpoints = [1, 10, 25, 50, 100, 250, 500, 1000, 2000, 3000]

print("\n--- Part D Checkpoints ---")
print("Epoch | Loss | Accuracy")
print("-" * 35)

for epoch in checkpoints:
    print(
        f"{epoch:5d} | "
        f"{loss_history_d[epoch - 1]:.4f} | "
        f"{accuracy_history_d[epoch - 1]:.4f}"
    )
```

```python
# -----------------------------
# Part E: Change the Part C dataset to make convergence slower
# -----------------------------

np.random.seed(42)

n_per_class = 40

# Slower-convergence dataset:
# Compared to Part C, the classes are closer to each other.
# The dataset is still learnable, but the margin between the classes is smaller.
class_0_e = np.random.normal(
    loc=[0.46, 0.46],
    scale=[0.03, 0.03],
    size=(n_per_class, 2)
)

class_1_e = np.random.normal(
    loc=[0.54, 0.54],
    scale=[0.03, 0.03],
    size=(n_per_class, 2)
)

X_slow = np.vstack([class_0_e, class_1_e])
X_slow = np.clip(X_slow, 0, 1)

y_slow = np.vstack([
    np.zeros((n_per_class, 1)),
    np.ones((n_per_class, 1))
])

dataset_e = pd.DataFrame(
    np.hstack([X_slow, y_slow]),
    columns=["x1_noise_level", "x2_vibration_level", "target"]
)

print("Slower-convergence synthetic dataset shape:", dataset_e.shape)
print("\nFirst 10 rows of the slower-convergence dataset:")
display(dataset_e.head(10))

print("\nClass distribution:")
print(dataset_e["target"].value_counts().sort_index())


# -----------------------------
# Plot Part E dataset
# -----------------------------

plt.figure(figsize=(7, 6))

plt.scatter(
    X_slow[y_slow.flatten() == 0, 0],
    X_slow[y_slow.flatten() == 0, 1],
    label="Class 0 - Normal"
)

plt.scatter(
    X_slow[y_slow.flatten() == 1, 0],
    X_slow[y_slow.flatten() == 1, 1],
    label="Class 1 - Abnormal"
)

plt.title("Slower-Convergence Synthetic Classification Dataset - Part E")
plt.xlabel("x1 - Noise Level")
plt.ylabel("x2 - Vibration Level")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Train the same 2-3-1 network
# -----------------------------

EPOCHS_E = 3000
LEARNING_RATE_E = 1.0

np.random.seed(7)

nn_e = NeuralNetwork231(input_size=2, hidden_size=3, output_size=1)

loss_history_e, accuracy_history_e = nn_e.train(
    X_slow,
    y_slow,
    epochs=EPOCHS_E,
    learning_rate=LEARNING_RATE_E
)

final_predictions_e = nn_e.forward(X_slow)
final_classes_e = (final_predictions_e >= 0.5).astype(int)

final_loss_e = nn_e.binary_cross_entropy(y_slow, final_predictions_e)
final_accuracy_e = np.mean(final_classes_e == y_slow)

print("\n--- Training Results For Part E ---")
print(f"Epochs: {EPOCHS_E}")
print(f"Learning rate: {LEARNING_RATE_E}")
print(f"Final loss: {final_loss_e:.4f}")
print(f"Final accuracy: {final_accuracy_e:.4f}")

print("\nFirst 10 predictions:")
for i in range(10):
    print(
        f"x = {X_slow[i]} | "
        f"prediction = {final_predictions_e[i][0]:.4f} | "
        f"class = {final_classes_e[i][0]} | "
        f"target = {int(y_slow[i][0])}"
    )


# -----------------------------
# Combined convergence plots
# -----------------------------

epochs_axis_c = np.arange(1, len(loss_history_c) + 1)
epochs_axis_e = np.arange(1, len(loss_history_e) + 1)

plt.figure(figsize=(10, 6))

plt.plot(epochs_axis_c, loss_history_c, linewidth=2, label="Part C Loss")

if "loss_history_d" in globals():
    epochs_axis_d = np.arange(1, len(loss_history_d) + 1)
    plt.plot(epochs_axis_d, loss_history_d, linewidth=2, label="Part D Loss")

plt.plot(epochs_axis_e, loss_history_e, linewidth=2, label="Part E Loss")

plt.title("Loss Comparison: Parts C, D, and E")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross Entropy Loss")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 6))

plt.plot(epochs_axis_c, accuracy_history_c, linewidth=2, label="Part C Accuracy")

if "accuracy_history_d" in globals():
    epochs_axis_d = np.arange(1, len(accuracy_history_d) + 1)
    plt.plot(epochs_axis_d, accuracy_history_d, linewidth=2, label="Part D Accuracy")

plt.plot(epochs_axis_e, accuracy_history_e, linewidth=2, label="Part E Accuracy")

plt.title("Accuracy Comparison: Parts C, D, and E")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()


# -----------------------------
# Convergence speed metrics
# -----------------------------

def first_epoch_at_or_below(values, threshold):
    values = np.array(values)
    indices = np.where(values <= threshold)[0]
    if len(indices) == 0:
        return None
    return int(indices[0] + 1)


def first_epoch_at_or_above(values, threshold):
    values = np.array(values)
    indices = np.where(values >= threshold)[0]
    if len(indices) == 0:
        return None
    return int(indices[0] + 1)


# Part C metrics
part_c_loss_below_020 = first_epoch_at_or_below(loss_history_c, 0.20)
part_c_loss_below_015 = first_epoch_at_or_below(loss_history_c, 0.15)
part_c_loss_below_010 = first_epoch_at_or_below(loss_history_c, 0.10)
part_c_loss_below_005 = first_epoch_at_or_below(loss_history_c, 0.05)
part_c_acc_above_090 = first_epoch_at_or_above(accuracy_history_c, 0.90)
part_c_acc_equal_100 = first_epoch_at_or_above(accuracy_history_c, 1.00)

# Part E metrics
part_e_loss_below_020 = first_epoch_at_or_below(loss_history_e, 0.20)
part_e_loss_below_015 = first_epoch_at_or_below(loss_history_e, 0.15)
part_e_loss_below_010 = first_epoch_at_or_below(loss_history_e, 0.10)
part_e_loss_below_005 = first_epoch_at_or_below(loss_history_e, 0.05)
part_e_acc_above_090 = first_epoch_at_or_above(accuracy_history_e, 0.90)
part_e_acc_equal_100 = first_epoch_at_or_above(accuracy_history_e, 1.00)

print("\n--- Convergence Speed Comparison ---")
print("Metric | Part C | Part D | Part E")
print("-" * 80)

if (
    "loss_history_d" in globals()
    and "accuracy_history_d" in globals()
    and "final_loss_d" in globals()
    and "final_accuracy_d" in globals()
):
    part_d_loss_below_020 = first_epoch_at_or_below(loss_history_d, 0.20)
    part_d_loss_below_015 = first_epoch_at_or_below(loss_history_d, 0.15)
    part_d_loss_below_010 = first_epoch_at_or_below(loss_history_d, 0.10)
    part_d_loss_below_005 = first_epoch_at_or_below(loss_history_d, 0.05)
    part_d_acc_above_090 = first_epoch_at_or_above(accuracy_history_d, 0.90)
    part_d_acc_equal_100 = first_epoch_at_or_above(accuracy_history_d, 1.00)

    print(f"Final loss | {final_loss_c:.4f} | {final_loss_d:.4f} | {final_loss_e:.4f}")
    print(f"Final accuracy | {final_accuracy_c:.4f} | {final_accuracy_d:.4f} | {final_accuracy_e:.4f}")
    print(f"First epoch with loss <= 0.20 | {part_c_loss_below_020} | {part_d_loss_below_020} | {part_e_loss_below_020}")
    print(f"First epoch with loss <= 0.15 | {part_c_loss_below_015} | {part_d_loss_below_015} | {part_e_loss_below_015}")
    print(f"First epoch with loss <= 0.10 | {part_c_loss_below_010} | {part_d_loss_below_010} | {part_e_loss_below_010}")
    print(f"First epoch with loss <= 0.05 | {part_c_loss_below_005} | {part_d_loss_below_005} | {part_e_loss_below_005}")
    print(f"First epoch with accuracy >= 0.90 | {part_c_acc_above_090} | {part_d_acc_above_090} | {part_e_acc_above_090}")
    print(f"First epoch with accuracy = 1.00 | {part_c_acc_equal_100} | {part_d_acc_equal_100} | {part_e_acc_equal_100}")

else:
    print(f"Final loss | {final_loss_c:.4f} | Not available | {final_loss_e:.4f}")
    print(f"Final accuracy | {final_accuracy_c:.4f} | Not available | {final_accuracy_e:.4f}")
    print(f"First epoch with loss <= 0.20 | {part_c_loss_below_020} | Not available | {part_e_loss_below_020}")
    print(f"First epoch with loss <= 0.15 | {part_c_loss_below_015} | Not available | {part_e_loss_below_015}")
    print(f"First epoch with loss <= 0.10 | {part_c_loss_below_010} | Not available | {part_e_loss_below_010}")
    print(f"First epoch with loss <= 0.05 | {part_c_loss_below_005} | Not available | {part_e_loss_below_005}")
    print(f"First epoch with accuracy >= 0.90 | {part_c_acc_above_090} | Not available | {part_e_acc_above_090}")
    print(f"First epoch with accuracy = 1.00 | {part_c_acc_equal_100} | Not available | {part_e_acc_equal_100}")


# -----------------------------
# Part E checkpoint table
# -----------------------------

checkpoints = [1, 10, 25, 50, 100, 250, 500, 1000, 2000, 3000]

print("\n--- Part E Checkpoints ---")
print("Epoch | Loss | Accuracy")
print("-" * 35)

for epoch in checkpoints:
    print(
        f"{epoch:5d} | "
        f"{loss_history_e[epoch - 1]:.4f} | "
        f"{accuracy_history_e[epoch - 1]:.4f}"
    )
```
